## 1. Objetivo do teste

Validar, com pessoas que não desenvolveram o sistema, dois aspectos:
- **Funcionamento:** o sistema sinaliza corretamente CONFORME / NÃO CONFORME conforme os EPIs em uso.
- **Usabilidade:** o quão fácil e claro é interagir com o sistema, e a percepção geral do usuário.


## 2. Perfil dos voluntários

- Pessoas externas à equipe (colegas, usuários em potencial), sem treinamento prévio no sistema.
- Meta de 6 a 10 participantes (realizado: 8).
- Participação voluntária; nenhum dado pessoal sensível é coletado ou armazenado — apenas as respostas do
  questionário (de forma anônima na análise) e a filmagem da interação, com consentimento verbal.


## 3. Materiais

- Computador com o sistema em execução (`deteccao_epi.py`) e a câmera **fixa e calibrada**.
- EPIs de teste: **capacete** e **colete de alta visibilidade**.
- Celular para filmar a interação.
- Questionário (Seção 6), um por voluntário.


## 4. Procedimento (passo a passo do avaliador)

1. Posicionar a câmera fixa cobrindo a área de aproximação; iniciar o sistema.
2. Explicar brevemente ao voluntário o cenário: "aproxime-se da câmera como se fosse entrar em uma área
   que exige EPI".
3. Pedir que o voluntário execute as **quatro situações**, observando o status exibido em cada uma:

| Situação | O que o voluntário faz | Status esperado |
|----------|------------------------|-----------------|
| A | Com **capacete e colete** | CONFORME |
| B | **Sem capacete** (só colete) | NÃO CONFORME (falta capacete) |
| C | **Sem colete** (só capacete) | NÃO CONFORME (falta colete) |
| D | **Sem nenhum EPI** | NÃO CONFORME |

4. Filmar a interação (celular).
5. Registrar observações do avaliador: acertos/erros do sistema em cada situação e qualquer falha.
6. Entregar o questionário (Seção 6) para o voluntário preencher.
7. Agradecer e coletar comentários finais.


## 5. O que registrar

- Resultado do sistema em cada situação (A–D) versus o esperado (para a análise de acertos).
- Dificuldades observadas durante o uso.
- Respostas do questionário (usabilidade + perguntas abertas).
- Sugestões e feedback espontâneo.


## 6. Instrumento de coleta (questionário aplicado)

**Identificação:** nome completo · nome do sistema · data.

**Parte A — Escala de usabilidade** (marque 1 a 5; 1 = discordo totalmente, 5 = concordo totalmente):

1. Gostaria de usar esse sistema com frequência.
2. Achei o sistema desnecessariamente complexo.
3. Achei o sistema fácil de usar.
4. Acho que precisaria de suporte de uma pessoa técnica para usar esse sistema.
5. Achei que várias funções do sistema foram bem integradas.
6. Achei que havia várias inconsistências no sistema.
7. Imagino que a maioria das pessoas aprenderia a usar esse sistema rapidamente.
8. Achei o sistema complicado de usar.
9. Me senti confiante ao usar o sistema.
10. Precisei aprender muitas coisas antes que pudesse usar o sistema.
11. Você achou o sistema interativo?

**Parte B — Perguntas abertas:**

12. Do que você mais gostou no sistema?
13. Do que você menos gostou no sistema?
14. Sugestões e comentários.
15. Qual é o objetivo deste sistema testado?
16. Descreva resumidamente os experimentos que você realizou com o sistema.
17. Descreva resumidamente os resultados obtidos nestes experimentos.

> A pontuação de usabilidade segue a escala **SUS (System Usability Scale)**: itens ímpares são positivos
> e itens pares são reversos; a nota final varia de 0 a 100 (acima de ~68 = boa usabilidade). O item 11 e
> as perguntas abertas complementam a avaliação.


## 7. Ética e consentimento

- Participação voluntária, com explicação prévia do objetivo do teste.
- Consentimento verbal para a filmagem; as imagens são usadas apenas para fins acadêmicos do trabalho.
- Na análise e nos relatórios, as respostas são tratadas de forma **anônima**.


## 8. Análise dos dados

Os questionários são transcritos e analisados em `Analise_Resultados_Voluntarios.ipynb` (cálculo do SUS,
consenso por item e síntese das respostas abertas). O confronto entre o resultado do sistema e o esperado
nas situações A–D alimenta a avaliação de acertos.
